```python
# INSERTION SORT

def insertion_sort(arr: list):
    
    n = len(arr)
    
    for i in range(1, n):
        
        key = arr[i]
        
        j = i - 1
        
        while j >= 0 and arr[j] > key:
            
            arr[j+1] = arr[j]
            
            j -=1
            
        
        arr[j+1] = key
    
    return arr
        
        
        
arr_unsorted = [ 31, 41, 59, 26, 41, 58 ]
arr_sorted = insertion_sort(arr=arr_unsorted)

print("Unsorted array: ", arr_unsorted)
print("Sorted array: ", arr_sorted)

```
